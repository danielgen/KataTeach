```python
# Cell 2: Create and show the board
# Adjust model_path to point to your model.ckpt file
board = create_interactive_board(
    model_path="model.ckpt",  # Update this path if needed
    board_size=19,
    device="auto",  # Will auto-detect mps/cuda/cpu
    prob_threshold=0.01  # Probability threshold for model moves
)

# Display the interactive board
# Click on intersections to play moves
# The model will automatically respond
board.show(figsize=(12, 12))
```

```python
# Cell 3 (optional): Get ownership values programmatically
# After playing some moves, you can access ownership:
ownership = board.get_ownership()
if ownership is not None:
    print(f"Ownership shape: {ownership.shape}")
    print(f"Ownership range: [{ownership.min():.3f}, {ownership.max():.3f}]")
    
    # Visualize ownership separately with proper scaling
    import matplotlib.pyplot as plt
    import numpy as np
    
    fig, ax = plt.subplots(figsize=(12, 12))
    
    # Calculate dynamic vmin/vmax based on actual data
    vmax = max(abs(ownership.max()), abs(ownership.min()), 0.01)
    vmin = -vmax
    
    # Display ownership
    im = ax.imshow(ownership, cmap='RdBu_r', vmin=vmin, vmax=vmax, origin='upper', interpolation='nearest')
    
    # Add colorbar
    cbar = plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
    cbar.set_label('Ownership (positive=current player, negative=opponent)', fontsize=12)
    
    # Add grid lines
    ax.set_xticks(np.arange(-0.5, 19, 1), minor=True)
    ax.set_yticks(np.arange(-0.5, 19, 1), minor=True)
    ax.grid(which='minor', color='black', linestyle='-', linewidth=0.5, alpha=0.3)
    
    # Add coordinate labels
    col_labels = 'ABCDEFGHJKLMNOPQRST'
    for i, label in enumerate(col_labels):
        ax.text(i, -0.5, label, ha='center', va='top', fontsize=8)
        ax.text(i, 18.5, label, ha='center', va='bottom', fontsize=8)
    
    for i in range(19):
        num_label = str(19 - i)
        ax.text(-0.5, i, num_label, ha='right', va='center', fontsize=8)
        ax.text(18.5, i, num_label, ha='left', va='center', fontsize=8)
    
    # Draw stones on top
    for y in range(19):
        for x in range(19):
            loc = board.gs.board.loc(x, y)
            stone = board.gs.board.board[loc]
            if stone == 1:  # Black
                circle = plt.Circle((x, y), 0.35, color='black', zorder=10)
                ax.add_patch(circle)
            elif stone == 2:  # White
                circle = plt.Circle((x, y), 0.35, color='white', edgecolor='black', linewidth=0.5, zorder=10)
                ax.add_patch(circle)
    
    # Add ownership value annotations for significant values
    for y in range(19):
        for x in range(19):
            own_val = ownership[y, x]
            if abs(own_val) > 0.15:  # Only show significant values
                val_str = f'{own_val:.1f}'
                text_color = 'white' if abs(own_val) > 0.3 else 'black'
                ax.text(x, y, val_str, ha='center', va='center', 
                       fontsize=7, color=text_color, weight='bold',
                       bbox=dict(boxstyle='round,pad=0.1', facecolor='white', alpha=0.7, edgecolor='none'),
                       zorder=15)
    
    ax.set_xlim(-0.5, 18.5)
    ax.set_ylim(18.5, -0.5)
    ax.set_title('Current Ownership Map', fontsize=14, fontweight='bold')
    ax.set_aspect('equal')
    
    plt.tight_layout()
    plt.show()
else:
    print("No ownership data available. Play some moves first!")
```

```python
# Cell 4 (optional): Get move history
moves = board.get_move_history()
print(f"Total moves played: {len(moves)}")
for i, (player, loc) in enumerate(moves):
    player_str = "Black" if player == 1 else "White"
    if loc == 0:  # Pass
        loc_str = "pass"
    else:
        # Convert to human-readable coordinates
        col_labels = 'ABCDEFGHJKLMNOPQRST'
        x = board.gs.board.loc_x(loc)
        y = board.gs.board.loc_y(loc)
        loc_str = f"{col_labels[x]}{19 - y}"
    print(f"Move {i+1}: {player_str} plays {loc_str}")
```
